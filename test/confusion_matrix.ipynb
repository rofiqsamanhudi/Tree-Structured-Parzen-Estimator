```python
from PIL import Image

# Path gambar
image_paths = [
    r"D:\Tree-Structured Parzen Estimator\main\xception\full_pipeline_results\confusion_matrix_inceptionv3.png",
    r"D:\Tree-Structured Parzen Estimator\main\vit\full_pipeline_results\confusion_matrix_vit.png",
    r"D:\Tree-Structured Parzen Estimator\main\inceptionv3\full_pipeline_results\confusion_matrix_inceptionv3.png",
    r"D:\Tree-Structured Parzen Estimator\main\densenet121\full_pipeline_results\confusion_matrix_densenet121.png"
]

# Buka gambar
images = [Image.open(path) for path in image_paths]

# Samakan ukuran gambar berdasarkan gambar pertama
target_size = images[0].size
images = [img.resize(target_size) for img in images]

# Ukuran canvas gabungan (2 kolom x 2 baris)
width, height = target_size
combined_image = Image.new(
    'RGB',
    (width * 2, height * 2),
    color='white'
)

# Tempel gambar
combined_image.paste(images[0], (0, 0))
combined_image.paste(images[1], (width, 0))
combined_image.paste(images[2], (0, height))
combined_image.paste(images[3], (width, height))

# Simpan hasil
output_path = r"D:\Tree-Structured Parzen Estimator\main\combined_confusion_matrix.png"
combined_image.save(output_path)

print(f"Gambar berhasil disimpan di:\n{output_path}")
```
